**Query Translation: Multi Query**

```python
from dotenv import load_dotenv
from openai import AsyncOpenAI

load_dotenv()
```

```python
client = AsyncOpenAI()
```

```python
async def generate_queries(question: str, n: int = 5) -> list[str]:
    SYSTEM_PROMPT = f"""You are an expert in query reformulation for document retrieval.

Given a user question generate {n} alternative queries using these strategies:
1. Rephrase with different vocabulary
2. Approach from a different perspective
3. Break into sub-questions if complex

Output only the queries, one per line, no numbering, no explanation, no "None"."""

    response = await client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {"role": "system", "content": SYSTEM_PROMPT},
            {"role": "user", "content": question},
        ],
        temperature=0.7,
    )

    raw = response.choices[0].message.content.strip()
    queries = [
        line.strip()
        for line in raw.split("\n")
        if line.strip() and line.strip().lower() != "none"
    ]

    return [question] + queries
```

```python
reformulated_queries = await generate_queries(question="what is fs module", n=3)
print(reformulated_queries)
```
