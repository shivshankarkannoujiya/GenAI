**RANKING**

`````python
def reciprocal_rank_fusion(rankings, k = 60):
    scores = {}
    for ranking in rankings:
        for rank, doc_id in enumerate(ranking):
            scores[doc_id] = scores.get(doc_id, 0) + 1 / (k + rank + 1)
    return sorted(scores.items(), key=lambda x: x[1], reverse=True)
`````

`````python
r1 = ["doc_a", "doc_b", "doc_c"]
r2 = ["doc_c", "doc_a", "doc_d"]
`````

`````python
print(reciprocal_rank_fusion([r1, r2]))
`````

## Reciprocal Rank Fusion (RRF) — Step by Step

```python
r1 = ["doc_a", "doc_b", "doc_c"]
r2 = ["doc_c", "doc_a", "doc_b"]

reciprocal_rank_fusion([r1, r2], k=60)
```

**Formula:** for each document $d$ across all rankings:

$$\text{score}(d) = \sum_{\text{rankings}} \frac{1}{k + \text{rank} + 1}$$

---

### Step 1 — Iterate through `r1`

| doc | rank | calculation | score |
|-----|------|-------------|-------|
| doc_a | 0 | $1 / (60 + 0 + 1)$ | 0.016393 |
| doc_b | 1 | $1 / (60 + 1 + 1)$ | 0.016129 |
| doc_c | 2 | $1 / (60 + 2 + 1)$ | 0.015873 |

---

### Step 2 — Iterate through `r2` and accumulate

| doc | rank | calculation | existing | new score |
|-----|------|-------------|----------|-----------|
| doc_c | 0 | $1 / (60 + 0 + 1)$ | 0.015873 | **0.032266** |
| doc_a | 1 | $1 / (60 + 1 + 1)$ | 0.016393 | **0.032522** |
| doc_b | 2 | $1 / (60 + 2 + 1)$ | 0.016129 | **0.032002** |

---

### Step 3 — Sort descending

| final rank | doc | score |
|------------|-----|-------|
| 1 | doc_a | 0.032522 |
| 2 | doc_c | 0.032266 |
| 3 | doc_b | 0.032002 |

---

### Intuition

- **doc_a** wins: ranked #1 in r1 and #2 in r2 — consistently near the top
- **doc_c** ranked #1 in r2 but only #3 in r1 — one strong rank wasn't enough
- **doc_b** ranked #2 in r1 and #3 in r2 — solid but never at the top

> RRF rewards **consistent high placement** across rankings over a single spectacular rank.
> The constant $k=60$ (Cormack et al., 2009) prevents top-ranked docs from dominating.
